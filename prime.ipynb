```python
#Basic code to check a number is prime
n = int(input("Enter the number: "))
prime = True
if n > 2:
    for i in range(2,n):
        if n % i == 0:
            prime = False
            break
        else:
            prime = True
else:
    print("Number is 1")

if prime:
    print(n,"is a prime number")
else:
    print(n,"is not a prime")
```
