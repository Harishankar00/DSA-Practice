```python
#Factorial without recursion
n = int(input("Enter a number for factorial:"))
fac = 1
for i in range(1,n+1):
    fac *= i
print(fac)
```

```python
#Palindrome Code
string1 = input("Enter a string:")
string2 =""
for i in string1:
    string2+=i.lower()
if string2 == string2[::-1]:
    print("The give string is a palindrome")
else:
    print("It is not a palindrome")
```

```python
#Prime Nummber code
n = int(input())
prime = True
if n > 2:
    for i in range(2,n):
        if n%i == 0:
            prime = False
            break
    if prime:
        print(n," is a prime number")
    else:
        print(n," is not a prime number")
elif n == 2:
    print("2 is a prime number")
else:
    print("The entered number is 1")
```

```python
#Armstrong number using String operation
n = int(input("Enter a number:"))
string = str(n)
l = len(str(n))
sums = 0
for i in range(l):
    sums += int(string[i])**l
if sums == n:
    print(n,"is an Armstrong number")
else:
    print(n,"is not an armstrong number")
```

```python
#Second Largest in a list
l1 = [4,7,3,12,76,89]
max = l1[0]
sec = l1[0]
for i in l1:
    if max < i:
        sec = max
        max = i
    elif i < max and sec < i:
        sec = i


print(sec,"is the second largest")
```

```python
#Finding missing values in an array
l = [7,9,1,2,4,5,6]
missing_values = []
big = l[0]
small = l[0]
for i in l:
    if i > big:
        big = i
    if i < small:
        small = i
for i in range(small,big+1):
    if i not in l:
        missing_values.append(i)
print("missing_values are",missing_values)
        
```

```python
#Most frequent Elements
l = [1,3,5,7,1,2,3,5,8,7,8,9,4,2,5,7,4,3,2,4,6,7,8,9,0,6,4,3]
dup_removed = []
val_count = []
for i in l:
    if i not in dup_removed:
        dup_removed.append(i)
for i in dup_removed:
    count = 0
    for j in range(len(l)):
        if i == l[j]:
            count+=1
    val_count.append(count)

max_val = 0
highest = 0
for i in range(len(val_count)):
    if max_val < val_count[i]:
        max_val = val_count[i]
for i in range(len(val_count)):
    if max_val == val_count[i]:
        highest = dup_removed[i]
print("The Highest Frequency number is",highest,"with frequency equal to",max_val)
```

```python
#Anagram Code
string1 = input("Enter a string1:")
string2 = input("Enter a string2:")
if len(string1) != len(string2):
    print(string1,"and",string2,"are not anagram")
else:
    freq1 = {}
    freq2 = {}
    for i in string1:
        count = 0
        if i not in freq1:
            for j in string1:
                if i == j:
                    count+=1
        freq1[i] = count
    for i in string2:
        count = 0
        if i not in freq2:
            for j in string2:
                if i == j:
                    count+=1
        freq2[i] = count    
    anagram_check = True
    for i in string1:
        if freq1[i] != freq2[i]:
            anagram_check = False
            break
        else:
            continue
    if anagram_check:
        print(string1,"and",string2,"are anagram")
    else:
        print(string1,"and",string2,"are not anagram")
    
```
