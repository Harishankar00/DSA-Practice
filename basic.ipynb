```python
#Greatest among 2 numbers
a = int(input("Enter a number:"))
b = int(input("Enter a number:"))
if a < b:
    print("B is Greater")
elif a == b:
    print("A , B is equal")
else:
    print("A is greater")
```

```python
#Maximum in an array
a = [ 4,6,7,2,4,1]
max = 0
for i in a:
    if i > max:
        max = i
print(max)
```

```python
#Swapping without temp
a = int(input("Enter a number:"))
b = int(input("Enter a number:"))
print("Before swap:" , a," ", b)
a , b = b , a
print("After swap:" , a, " ", b)
```

```python
#Factorial without recursion
n = int(input("Enter a number for factorial:"))
fac = 1
for i in range(1,n+1):
    fac *= i
print(fac)
```

```python
#Vowels count
a = ['a','e','i','o','u']
count = 0
string = input("Enter a string")
for i in string:
    if i in a:
        count+=1
print("Vowels:",count)
```

```python
#Palindrome Code
string1 = input("Enter a string:")
string2 =""
for i in string1:
    string2+=i.lower()
if string2 == string2[::-1]:
    print("The give string is a palindrome")
else:
    print("It is not a palindrome")
```

```python
#Prime Nummber code
n = int(input())
prime = True
if n > 2:
    for i in range(2,n):
        if n%i == 0:
            prime = False
            break
    if prime:
        print(n," is a prime number")
    else:
        print(n," is not a prime number")
elif n == 2:
    print("2 is a prime number")
else:
    print("The entered number is 1")
```

```python
#Armstrong number using String operation
n = int(input("Enter a number:"))
string = str(n)
l = len(str(n))
sums = 0
for i in range(l):
    sums += int(string[i])**l
if sums == n:
    print(n,"is an Armstrong number")
else:
    print(n,"is not an armstrong number")
```

```python
#Second Largest in a list
l1 = [4,7,3,12,76,89]
max = l1[0]
sec = l1[0]
for i in l1:
    if max < i:
        sec = max
        max = i
    elif i < max and sec < i:
        sec = i


print(sec,"is the second largest")
```

```python
#Finding missing values in an array
l = [7,9,1,2,4,5,6]
missing_values = []
big = l[0]
small = l[0]
for i in l:
    if i > big:
        big = i
    if i < small:
        small = i
for i in range(small,big+1):
    if i not in l:
        missing_values.append(i)
print("missing_values are",missing_values)
        
```
